```python
# 모델 초기화
model = YOLO('yolov8l.pt')  # nano 모델 사용 (n, s, m, l, x 중 선택)
```

```python
results = model.train( # 'el' 대신 'model' 사용
    data='/content/dataset/data.yaml',  # 데이터셋 설정 파일
    epochs=50,               # 학습 에폭 수
    imgsz=640,               # 이미지 크기
    batch=16,                # 배치 크기
    name='yolov8_custom',    # 프로젝트 이름
    project='/content/runs/train',    # 저장 경로
    patience=10,             # Early stopping patience
    save=True,               # 체크포인트 저장
    device=0                 # GPU 사용
)
```
