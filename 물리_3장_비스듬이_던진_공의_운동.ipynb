```python

import math

# 0. 변수 선언
PI = 3.141592
g = 9.8  # 중력 가속도 (m/s^2)

# 사용자 입력
theta_deg = float(input("각도를 입력하세요 ==> "))
v0 = float(input("초기속도를 입력하세요 ==> "))
y0 = float(input("초기높이를 입력하세요 ==> "))

# 라디안으로 변환
theta_rad = theta_deg * PI / 180.0

# 1. 초기 속도 성분 분해
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)

# 2차 방정식의 계수 설정 (높이 방정식: y = y0 + v0y*t - 0.5*g*t^2)
a = -0.5 * g
b = v0y
c = y0
D = b**2 - 4*a*c

if D < 0:
    print("오류: 실근이 존재하지 않습니다.")
    exit(1)

# 지면에 도달할 때까지의 시간 (양수인 근 선택)
t1 = (-b + math.sqrt(D)) / (2 * a)
t2 = (-b - math.sqrt(D)) / (2 * a)
t3 = max(t1, t2)

print(f"(a) 지면에 도달할 때까지의 시간: {t3:.2f} 초")

# (b) 수평 거리 계산
dh = v0x * t3
print(f"(b) 수평 거리: {dh:.2f} m")

# (c) 속도의 수평 및 수직 성분 계산
vx = v0x
vy = v0y - g * t3
print(f"(c) 속도의 수평 성분: {vx:.2f} m/s")
print(f"(c) 속도의 수직 성분: {vy:.2f} m/s")

# (d) 속도의 크기 계산
s = math.sqrt(vx**2 + vy**2)
print(f"(d) 속도의 크기: {s:.2f} m/s")

# (e) 속도 벡터와 수평 방향이 이루는 각도 계산
ang = math.degrees(math.atan2(vy, vx))
print(f"(e) 수평 방향과 이루는 각도: {ang:.2f} 도")

```
